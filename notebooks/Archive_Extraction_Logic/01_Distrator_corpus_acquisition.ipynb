## Initial Setup and Environment Preparation




Install the arxiv library: This step is crucial for interacting with the arXiv API. The arxiv Python client library allows us to programmatically search for scientific papers, filter them based on specific keywords (like 'density functional theory' and 'benchmark'), and download their PDF versions directly from arXiv. This automation is essential for efficiently gathering a large dataset of research papers without manual intervention.

Create the data/distractors directory: This directory serves as an organized storage location for the PDF files downloaded from arXiv. By creating a dedicated directory, we ensure that the downloaded papers are kept separate from other project files, making them easy to locate, manage, and process later. The naming convention (distractors) hints at their role in a larger context, likely as a 'haystack' of irrelevant documents from which specific 'needles' (target documents) need to be found in a retrieval task.

```python
import sys
!{sys.executable} -m pip install arxiv

import arxiv
import os

# 1. Setup
save_dir = "data/distractors"
if not os.path.exists(save_dir):
    os.makedirs(save_dir)

# 2. Define the Search (Broad DFT terms)
# We exclude "review" to get research papers, and sort by submitted date to get random recent ones
client = arxiv.Client()
search = arxiv.Search(
    query = 'abs:"density functional theory" AND abs:"benchmark"',
    max_results = 80, # Get a few extra in case of errors
    sort_by = arxiv.SortCriterion.SubmittedDate
)

print(f"Starting download of distractor papers into '{save_dir}'...")

count = 0
for result in client.results(search):
    if count >= 75:
        break

    try:
        # 3. Construct a clean filename (Distractor_01.pdf)
        # We rename them immediately so you NEVER confuse them with your Gold Set
        filename = f"Distractor_{count+1:03d}.pdf"
        file_path = os.path.join(save_dir, filename)

        # 4. Download
        result.download_pdf(dirpath=save_dir, filename=filename)
        print(f"[{count+1}/75] Downloaded: {filename}")
        count += 1

    except Exception as e:
        print(f"Skipped one due to error: {e}")

print("Done! You now have your Haystack.")
```

## Archiving and Downloading Distractor Papers


Archive the downloaded distractor papers into a zip file and provide it for download.


```python
import shutil
from google.colab import files
import os

target_dir = "data/distractors"
zip_filename = 'distractors.zip'

if os.path.exists(target_dir) and os.listdir(target_dir):
    # Zip the directory
    shutil.make_archive(zip_filename.split('.')[0], 'zip', target_dir)

    # Download the zip file
    files.download(zip_filename)
else:
    print(f"Error: Directory '{target_dir}' not found or is empty. Please ensure papers are downloaded by running the previous cell (U-836SmMwU9F) first.")
```
